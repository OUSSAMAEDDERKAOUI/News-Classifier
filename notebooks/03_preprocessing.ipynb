```python
import pandas as pd
df_train= pd.read_csv("./data/raw/df_train.csv")
df_test=pd.read_csv("./data/raw/df_test.csv")
```

```python
import pandas as pd
import re

def normalize_text(text):
    text = text.lower()  
    text = re.sub(r'\s+', ' ', text)  
    return text.strip()


df_train['text'] = df_train['text'].apply(normalize_text)

df_test['text'] = df_test['text'].apply(normalize_text)
```
