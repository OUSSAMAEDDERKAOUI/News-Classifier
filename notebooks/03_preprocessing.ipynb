```python
import pandas as pd
df_train= pd.read_csv("./data/raw/df_train.csv")
df_test=pd.read_csv("./data/raw/df_test.csv")
```

```python
import pandas as pd
import re

def normalize_text(text):
    text = text.lower()  
    text = re.sub(r'\s+', ' ', text)  
    return text.strip()


df_train['text'] = df_train['text'].apply(normalize_text)

df_test['text'] = df_test['text'].apply(normalize_text)
```

```python
print("Nombre de doublons dans df_train :", df_train.duplicated().sum())
print("Nombre de doublons dans df_test :", df_test.duplicated().sum())

df_train = df_train.drop_duplicates()
df_test = df_test.drop_duplicates()

```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.data.path.append('/root/nltk_data')

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    tokens = word_tokenize(text)
    filtered_tokens = [w for w in tokens if w.lower() not in stop_words]
    return " ".join(filtered_tokens)

df_train['text_clean'] = df_train['text'].apply(remove_stopwords)
df_test['text_clean']  = df_test['text'].apply(remove_stopwords)
```

```python
def remove_punctuation(text):
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)  
    text = re.sub(r'\s+', ' ', text)  
    return text.strip()

df_train['text_clean'] = df_train['text'].apply(remove_punctuation)
df_test['text_clean'] = df_test['text'].apply(remove_punctuation)
```

```python
df_train.to_csv('./data/processed/df_train_preprocessed.csv',index=False)
display(df_train)
```
